## 5. Pandas analysis
Create at least two calculated columns or useful summaries. Filter and sort rows to answer questions about your dataset.


```python
# 1. Create calculated columns

df['Total_Points'] = df['Running_Points'] + df['Jump_Points'] + df['Throw_Points']

df['Average_Points'] = df[['Running_Points', 'Jump_Points', 'Throw_Points']].mean(axis=1)

print("Calculated Columns:")
print(df.head())


# 2. Filter records using a condition

high_performers = df[df['Total_Points'] > 50]

print("\nStudents scoring more than 50 points:")
print(high_performers)


# 3. Sort records and display top/bottom results

sorted_students = df.sort_values(by='Total_Points', ascending=False)

print("\nTop 5 Students:")
print(sorted_students.head(5))

print("\nBottom 5 Students:")
print(sorted_students.tail(5))
```

## 6. Your individual project challenge
**Challenge:** Calculate total points; rank participants; identify strongest event by average points.

Write your solution below and explain what the result means.


```python
# Calculate total points

df['Total_Points'] = df['Running_Points'] + df['Jump_Points'] + df['Throw_Points']

print("Total Points:")
print(df[['Name', 'Total_Points']])


# Rank participants

df['Rank'] = df['Total_Points'].rank(ascending=False, method='min').astype(int)

print("\nStudent Ranking:")
print(df[['Name', 'Total_Points', 'Rank']].sort_values(by='Rank'))


# Find strongest event by average points

event_avg = df[['Running_Points', 'Jump_Points', 'Throw_Points']].mean()

print("\nAverage Points of Each Event:")
print(event_avg)

strongest_event = event_avg.idxmax()

print("\nStrongest Event:", strongest_event)
```

```python
#  Explanation

# In this project, I calculated the total points of each student by adding their Running, Jump and Throw points.

# After that, I ranked all participants according to their total points. The student with the highest total points got the first rank.

# I also calculated the average points of all three events to find which event had the best overall performance.

# The event with the highest average points is considered the strongest event in this dataset.
```
